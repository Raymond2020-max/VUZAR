## 2️⃣ Start VUZA Server

This will start the server and give you a **public URL** to access VUZA from any browser.

> 🎯 The T4 GPU makes video processing **MUCH faster**!
> You also get Google's fast internet for downloading media.

```
import subprocess
import threading
import time
from pyngrok import ngrok

# Start the VUZA server in background
server = subprocess.Popen(['python', 'app.py'], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
time.sleep(3)

# Create public URL
public_url = ngrok.connect(8000)
print(f"\n🚀 VUZA is LIVE!")
print(f"\n🌐 Open this URL in your browser:")
print(f"\n   👉 {public_url}")
print(f"\n📌 Keep this cell running! Close the tab to stop.")
print(f"\n💡 TIP: Add your API keys in the ⚙️ Settings panel")

# Keep alive
try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    ngrok.kill()
    server.terminate()
    print("\n🛑 Server stopped.")
```
